### Overall Recommender App Run

##### Step 1: Read Nutrient Values from Map (Coords selected by user)

- Import maps

```python
import rasterio
import os
dir = str(os.getcwd())
k_map = rasterio.open(os.path.join(dir, "static", "images", "k_map.tif"))
mg_map = rasterio.open(os.path.join(dir, "static", "images", "mg_map.tif"))
print(k_map.crs)
print(mg_map.crs)
print(k_map.bounds)
print(mg_map.bounds)
```

- Read ExK and ExMg from set coords

```python
y = -0.7306568885149762
x = 36.35131828787092
mg_row,mg_col = mg_map.index(x,y)
k_row, k_col = k_map.index(x,y)
exmg = mg_map.read(1)[mg_row, mg_col]
exk = k_map.read(1)[k_row, k_col]

print("ExMg value on long/lat [",x, y,"]:",exmg)
print("ExK value on long/lat [",x, y,"]:",exk,"\n")
```

- Convert to available soil K and Mg values

```python
conv_mg = 122
conv_k = 390

soil_mg = exmg * conv_mg
soil_k = exk * conv_k

print("Available K: ", round(soil_k,2), " Available Mg: ", round(soil_mg,2))
```
